# Analyze Traces

```python
import pandas as pd
import ast

# read in dataset of traces
trace_dataset_name = "EscapeRoom_v2"
seed = 1

df_train = pd.read_csv(f'methods/llm_as_judge/data/{trace_dataset_name}_train_df_seed{seed}.csv')
df_test = pd.read_csv(f'methods/llm_as_judge/data/{trace_dataset_name}_test_df_seed{seed}.csv')

df_train["step_trace"] = df_train["step_trace"].apply(ast.literal_eval)
df_test["step_trace"] = df_test["step_trace"].apply(ast.literal_eval)

# combine datasets
df = pd.concat([df_train, df_test], ignore_index=True)
```

```python
df.groupby("action_id")["step_label"].mean()
```
